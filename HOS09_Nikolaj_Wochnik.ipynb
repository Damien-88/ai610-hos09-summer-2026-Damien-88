```python
# Import Libraries
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Sample data
data = {
    "age": [35, 42, 28, 19, 56, 40, 33],
    "income": [60000, 80000, 45000, 25000, 100000, 75000, 55000],
    "education_level": [
        "High School", "Bachelor", "High School", "Associate", "Master", 
        "Bachelor", "Associate"
    ],
    "employment_status": [
        "Employed", "Employed", "Unemployed", "Student", "Employed", 
        "Employed", "Unemployed"
    ],
    "loan_approved": ["Yes", "Yes", "No", "No", "Yes", "Yes", "No"]
}

# Create DataFrame
df = pd.DataFrame(data)

# Select numerical columns for scaling
numerical_cols = ["age", "income"]

# Perform scaling on numerical columns
scaler = StandardScaler()
df_scaled = df.copy()
df_scaled[numerical_cols] = scaler.fit_transform(df[numerical_cols])

# Perform one-hot encoding on categorical columns
df_encoded = pd.get_dummies(
    df_scaled, 
    columns=["education_level", "employment_status", "loan_approved"], 
    drop_first = True, 
    dummy_na = True
)
# Covert Booleans to 1 or 0
df_encoded = df_encoded.astype(int)

# Separate features (X) and target (y)
X = df_encoded.drop(["loan_approved_Yes"], axis = 1).values
y = df_encoded["loan_approved_Yes"].values.reshape(-1, 1)

# Define the activation function (sigmoid)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Initialize the parameters of the neural network
input_dim = X.shape[1]
hidden_dim = 5
output_dim = 1
learning_rate = 0.1
epochs = 1000

# Initialize the weights and biases
np.random.seed(42)
W1 = np.random.randn(input_dim, hidden_dim)
b1 = np.zeros((1, hidden_dim))
W2 = np.random.randn(hidden_dim, output_dim)
b2 = np.zeros((1, output_dim))

# Training the neural network
for epoch in range(epochs):
    # Forward propagation
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    y_pred = sigmoid(z2)
    
    # Calculate the loss
    loss = -np.mean(y * np.log(y_pred) + (1 - y) * np.log(1 - y_pred))
    
    # Backpropagation
    delta2 = y_pred - y
    delta1 = np.dot(delta2, W2.T) * (a1 * (1 - a1))
    
    # Update the weights and biases
    W2 -= learning_rate * np.dot(a1.T, delta2)
    b2 -= learning_rate * np.sum(delta2, axis = 0, keepdims = True)
    W1 -= learning_rate * np.dot(X.T, delta1)
    b1 -= learning_rate * np.sum(delta1, axis = 0)
    
    # Print the loss for every 100 epochs
    if epoch % 100 == 0:
        print("Epoch:", epoch, "Loss:", loss)    
        
# Predict the classes
predictions = (y_pred > 0.5).astype(int)
print("Predictions:", predictions.flatten())

# Calculate accuracy
accuracy = np.mean(predictions.flatten() == y.flatten())
print("Accuracy:", accuracy)
```
